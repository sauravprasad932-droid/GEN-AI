```python

from langchain_ollama import ChatOllama

# Connect LangChain with Ollama
llm = ChatOllama(
    model="gemma3:1b",
    temperature=0.7,
    streaming=True
)
```

```python
question="can you explain me GEN-AI ?"
```

```python
response=llm.stream(question)
```
