```python

from langchain_ollama import ChatOllama

# Connect LangChain with Ollama
llm = ChatOllama(
    model="qwen2.5:3b",
    temperature=0.7,
    streaming=True
)
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
    it will return the sum of two numbers
    Args:
    a:Number One
    b:Number Two
    
    """
    return a+b

@tool
def mull_numbers(a:int,b:int):
    """
    it will return the product of two numbers
    Args:
    a:Number One
    b:Number Two
    
    """
    return a*b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[add_numbers, mull_numbers],
    system_prompt="You are a math teacher, and always use a tool for calculation."
)

response = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is 2+3?"}
    ]
})

print(response["messages"][-1].content)
```
