```python
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from dotenv import load_dotenv
import os

# Load API key
load_dotenv()

# Create LLM object
llm = ChatNVIDIA(
    model="nvidia/nemotron-3.5-bolt-e4b",
    nvidia_api_key=os.getenv("NVIDIA_API_KEY"),
    temperature=1.0,
    top_p=0.95
)

print("LLM created successfully!")
```

```python
response = llm.invoke(
    "are you free to use as api"
)

print(response.content)
```

```python
# STATIC PROMPTS
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

messages = [
  SystemMessage(content="you are a food lover from india"),
  HumanMessage(content="give me 5 best food of bihar"),
  AIMessage(content="use fish curry sarso "),
  HumanMessage(content="if fish currry sarso is in top 5 then give me recpie of how to make it else say no it is not present")
   
]

response = llm.invoke(messages)

print(response.content)
```

```python
messages = [
    SystemMessage(content="You are a Python developer."),

    HumanMessage(content="How do I sort an array?")
]

# First AI response
response1 = llm.invoke(messages)

print(response1.content)

# Store actual AI response in conversation history
messages.append(response1)

# Ask second question
messages.append(
    HumanMessage(content="Give me an example.")
)

# Second AI response
response2 = llm.invoke(messages)

print(response2.content)
```
