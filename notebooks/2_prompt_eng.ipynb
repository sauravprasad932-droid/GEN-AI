```python
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from dotenv import load_dotenv
import os

# Load API key
load_dotenv()

# Create LLM object
llm = ChatNVIDIA(
    model="nvidia/nemotron-3.5-bolt-e4b",
    nvidia_api_key=os.getenv("NVIDIA_API_KEY"),
    temperature=1.0,
    top_p=0.95
)

print("LLM created successfully!")
```

```python
prompts=[
    {"role":"system","content":"you are python devloper"},
    {"role":"human","content":"explain dictionary in hindi"}
    
]
result=llm.invoke(prompts)
print(result.content)
```

```python
# DYNAMIC-PROMPTING
from langchain_core.prompts import ChatPromptTemplate

prompts=ChatPromptTemplate.from_messages(
    [
        {"role":"system","content":"you are a translator and translate input in {language}"},
        {"role":"human","content":"{query}"}
         
    ]
)

final_prompts=prompts.format_messages(language="Hinglish",query="I love python")

result=llm.invoke(final_prompts)
print(result.content)
```
